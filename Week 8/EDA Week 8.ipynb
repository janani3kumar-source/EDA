```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cleaned StudentsPerformance.csv")
```

```python
df.head()
```

```python
scores = ["math score", "reading score", "writing score"]
```

```python
parental_avg = df.groupby("parental level of education")[scores].mean()

print(parental_avg)
```

```python
parental_avg.plot(kind="bar", figsize=(12, 6))

plt.title("Average Scores Based on Parental Level of Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.legend(title="Subjects")
plt.show()
```

```python
lunch_avg = df.groupby("lunch")[scores].mean()

print(lunch_avg)
```

```python
lunch_avg.plot(kind="bar", figsize=(10, 6))

plt.title("Average Scores Based on Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x="test preparation course", y="math score", data=df)

plt.title("Math Scores Based on Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Math Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x="test preparation course", y="reading score", data=df)

plt.title("Reading Scores Based on Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x="test preparation course", y="writing score", data=df)

plt.title("Writing Scores Based on Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Writing Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x="math score", y="reading score", data=df)

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x="reading score", y="writing score", data=df)

plt.title("Reading Score vs Writing Score")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x="reading score", y="writing score", data=df)

plt.title("Reading Score vs Writing Score")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
correlation_data = df[scores]

correlation_matrix = correlation_data.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap of Student Scores")
plt.show()
```

```python
print("Correlation between Reading and Writing:")
print(correlation_matrix.loc["reading score", "writing score"])
```
